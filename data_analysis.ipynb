```python
# Section 3.4 table (cost/latency/tokens) from dataset
from pathlib import Path
import json
import numpy as np
import pandas as pd

V1_PRICING = {"input": 1.25, "cached_input": 0.125, "output": 10.0, "web_search": 0.01}
V2_PRICING = {"input": 2.50, "cached_input": 0.25, "output": 15.0, "web_search": 0.01}


def _safe_num(x, default=0.0):
    try:
        if x is None:
            return default
        return float(x)
    except Exception:
        return default


def _summary_stats(values):
    arr = [float(x) for x in values if x is not None and not pd.isna(x)]
    if not arr:
        return {"mean": np.nan, "median": np.nan, "max": np.nan}
    return {
        "mean": float(np.mean(arr)),
        "median": float(np.median(arr)),
        "max": float(np.max(arr)),
    }


def load_v1_claim_metrics(dataset_root=Path('dataset')):
    rows = []
    for p in sorted((dataset_root / 'AIv1').glob('ai*.log')):
        d = json.loads(p.read_text(encoding='utf-8'))
        dr = d.get('deep_research_log', {}) or {}
        usage = dr.get('usage', {}) or {}
        input_tokens = int(_safe_num(usage.get('input_tokens'), 0))
        cached_tokens = int(_safe_num((usage.get('input_tokens_details') or {}).get('cached_tokens'), 0))
        output_tokens = int(_safe_num(usage.get('output_tokens'), 0))

        # v1 web searches in raw Responses API output
        output_items = dr.get('output') or []
        web_searches = int(sum(1 for item in output_items if (item or {}).get('type') == 'web_search_call'))

        created_at = _safe_num(dr.get('created_at'), np.nan)
        completed_at = _safe_num(dr.get('completed_at'), np.nan)
        wall_seconds = (completed_at - created_at) if (not np.isnan(created_at) and not np.isnan(completed_at)) else np.nan

        non_cached_input = max(0, input_tokens - cached_tokens)
        # Research call cost
        cost = (
            non_cached_input * V1_PRICING['input']
            + cached_tokens * V1_PRICING['cached_input']
            + output_tokens * V1_PRICING['output']
        ) / 1_000_000 + web_searches * V1_PRICING['web_search']
        # Approximate style post-processing overhead (as in report methodology)
        cost += (12_000 * V1_PRICING['input'] + 3_000 * V1_PRICING['output']) / 1_000_000

        rows.append({
            'claim_file': p.name,
            'wall_seconds': wall_seconds,
            'cost_usd': cost,
            'input_tokens': input_tokens,
            'cached_input_tokens': cached_tokens,
            'output_tokens': output_tokens,
            'web_searches': web_searches,
        })
    return pd.DataFrame(rows)


def load_v2_claim_metrics(dataset_root=Path('dataset')):
    rows = []
    for p in sorted((dataset_root / 'AIv2').glob('ai*.log')):
        d = json.loads(p.read_text(encoding='utf-8'))
        step_logs = d.get('step_logs') or []
        if not step_logs:
            continue
        steps = [s.get('content', {}) or {} for s in step_logs]
        last = steps[-1]

        input_tokens = int(_safe_num(last.get('input_tokens'), 0))
        cached_tokens = int(_safe_num(last.get('cached_input_tokens'), 0))
        output_tokens = int(_safe_num(last.get('output_tokens'), 0))
        web_searches = int(sum(int(_safe_num(s.get('web_searches'), 0)) for s in steps))
        wall_seconds = float(sum(_safe_num(s.get('elapsed_seconds'), 0) for s in steps))

        non_cached_input = max(0, input_tokens - cached_tokens)
        cost = (
            non_cached_input * V2_PRICING['input']
            + cached_tokens * V2_PRICING['cached_input']
            + output_tokens * V2_PRICING['output']
        ) / 1_000_000 + web_searches * V2_PRICING['web_search']

        rows.append({
            'claim_file': p.name,
            'wall_seconds': wall_seconds,
            'cost_usd': cost,
            'input_tokens': input_tokens,
            'cached_input_tokens': cached_tokens,
            'output_tokens': output_tokens,
            'web_searches': web_searches,
        })
    return pd.DataFrame(rows)


def build_section_3_4_table(v1_df, v2_df):
    v1_wall = _summary_stats(v1_df['wall_seconds'])
    v2_wall = _summary_stats(v2_df['wall_seconds'])
    v1_cost = _summary_stats(v1_df['cost_usd'])
    v2_cost = _summary_stats(v2_df['cost_usd'])

    rows = [
        ['Wall time, mean', f"{v1_wall['mean']:.0f} s", f"{v2_wall['mean']:.0f} s"],
        ['Wall time, median', f"{v1_wall['median']:.0f} s", f"{v2_wall['median']:.0f} s"],
        ['Wall time, max', f"{v1_wall['max']:.0f} s", f"{v2_wall['max']:.0f} s"],
        ['Cost per claim, mean', f"${v1_cost['mean']:.2f}", f"${v2_cost['mean']:.2f}"],
        ['Cost per claim, median', f"${v1_cost['median']:.2f}", f"${v2_cost['median']:.2f}"],
        ['Cost per claim, max', f"${v1_cost['max']:.2f}", f"${v2_cost['max']:.2f}"],
        ['Input tokens, mean', f"{v1_df['input_tokens'].mean()/1000:.0f} k", f"{v2_df['input_tokens'].mean()/1000:.0f} k"],
        ['Cached input tokens, mean', f"{v1_df['cached_input_tokens'].mean()/1000:.0f} k", f"{v2_df['cached_input_tokens'].mean()/1000:.0f} k"],
        ['Output tokens, mean', f"{v1_df['output_tokens'].mean()/1000:.0f} k", f"{v2_df['output_tokens'].mean()/1000:.0f} k"],
        ['Web searches, mean', f"{v1_df['web_searches'].mean():.0f}", f"{v2_df['web_searches'].mean():.0f}"],
    ]
    return pd.DataFrame(rows, columns=['Metric', 'v1 (gpt-5.1)', 'v2 (gpt-5.4 via Codex)'])


v1_metrics = load_v1_claim_metrics(Path('dataset'))
v2_metrics = load_v2_claim_metrics(Path('dataset'))
section_3_4_table = build_section_3_4_table(v1_metrics, v2_metrics)

print(f"v1 claims used: {len(v1_metrics)}")
print(f"v2 claims used: {len(v2_metrics)}")
section_3_4_table


```

```python
# Section 6.1 (IAA) recomputation from dataset/checklist.csv
import numpy as np
import pandas as pd


def cohen_kappa_binary(a, b):
    a = np.asarray(a, dtype=float)
    b = np.asarray(b, dtype=float)
    mask = ~(np.isnan(a) | np.isnan(b))
    a = a[mask].astype(int)
    b = b[mask].astype(int)
    n = len(a)
    if n == 0:
        return None, None, 0

    observed = float((a == b).mean())
    p1 = float((a == 1).mean())
    p0 = 1.0 - p1
    q1 = float((b == 1).mean())
    q0 = 1.0 - q1
    expected = p1 * q1 + p0 * q0

    # Degenerate marginals => kappa undefined
    if np.isclose(1.0 - expected, 0.0):
        return None, observed, n

    kappa = (observed - expected) / (1.0 - expected)
    return float(kappa), observed, n


checklist = pd.read_csv('dataset/checklist.csv')

# AIdev/old split used for inter-annotator agreement in the report
old = checklist[checklist['split'] == 'old'].copy()
claims_with_both = (
    old.groupby('claim_id')['annotator']
    .nunique()
    .pipe(lambda s: s[s == 2].index.tolist())
)
old = old[old['claim_id'].isin(claims_with_both)].copy()
annotators = sorted(old['annotator'].dropna().unique().tolist())

question_cols = [c for c in checklist.columns if c not in ['claim_id', 'split', 'annotator']]

# Keep only binary (0/1) rubric columns
binary_cols = []
for col in question_cols:
    vals = pd.to_numeric(checklist[col], errors='coerce').dropna()
    if len(vals) == 0:
        continue
    unique_vals = set(vals.unique().tolist())
    if unique_vals.issubset({0, 1}):
        binary_cols.append(col)

rows = []
for col in binary_cols:
    piv = old.pivot_table(index='claim_id', columns='annotator', values=col, aggfunc='first')
    if not all(a in piv.columns for a in annotators):
        continue

    a = pd.to_numeric(piv[annotators[0]], errors='coerce').values
    b = pd.to_numeric(piv[annotators[1]], errors='coerce').values
    kappa, observed, n_pairs = cohen_kappa_binary(a, b)

    # Report criterion: at least 5 paired observations and defined kappa
    if n_pairs >= 5 and kappa is not None:
        rows.append({
            'question': col,
            'kappa': kappa,
            'observed_agreement': observed,
            'n': n_pairs,
        })

iaa_df = pd.DataFrame(rows)
iaa_df = iaa_df.sort_values('kappa', ascending=False).reset_index(drop=True)

iaa_summary = {
    'doubly_annotated_claims': len(claims_with_both),
    'binary_questions_with_valid_kappa': int(len(iaa_df)),
    'mean_kappa': float(iaa_df['kappa'].mean()),
    'median_kappa': float(iaa_df['kappa'].median()),
    'min_kappa': float(iaa_df['kappa'].min()),
    'max_kappa': float(iaa_df['kappa'].max()),
    'mean_observed_agreement': float(iaa_df['observed_agreement'].mean()),
}

print('IAA summary (for report §6.1)')
for k, v in iaa_summary.items():
    if isinstance(v, float):
        print(f'- {k}: {v:.3f}')
    else:
        print(f'- {k}: {v}')

print()
print('Top 5 κ questions')
display(iaa_df[['question', 'kappa', 'observed_agreement', 'n']].head(5))

print('Bottom 5 κ questions')
display(iaa_df[['question', 'kappa', 'observed_agreement', 'n']].tail(5).sort_values('kappa'))


```
